```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

#import file into pandas
df = pd.read_csv('Jamboree_Admission.txt')
print(df.head())
```

```python
# Create a new dataframe for analysis of datatypes
df_analysis = pd.DataFrame(df.dtypes)

# Assign column header to dataframe
df_analysis.columns = ['DataType']

# Assign a column header to first column of dataframe
#df_analysis.reset_index(inplace=True)
#df_analysis.columns = ['Attribute', 'DataType']

df_analysis
```

```python
# Add a new columns Min, Max, Missing values to the dataframe_analysis for each feature
  # df1 = df.select_dtypes(include=np.number)
  # df['new_col'] = df['col'].apply(lambda x: x * 2).

# df_analysis['Min'] = df.min()
# df_analysis['Max'] = df.max()
df_analysis['MissingValues'] = df.isnull().sum()
df_analysis
```

```python
# Univariate Analysis (distribution plots of all the continuous variable(s) barplots/countplots of all the categorical variables in dataframe df

#GRE Score
GRE_Dist = sns.histplot(df['GRE Score'])
plt.title('GRE Score Distribution')
plt.xlabel('GRE Score')
plt.ylabel('Frequency')
plt.show(GRE_Dist)

#TOEFL Score
TOEFL_Dist = sns.histplot(df['TOEFL Score'])
plt.title('TOEFL Score Distribution')
plt.xlabel('TOEFL Score')
plt.ylabel('Frequency')
plt.show(TOEFL_Dist)

#University Rating
University_Rating_Dist = sns.histplot(df['University Rating'])
plt.title('University Rating Distribution')
plt.xlabel('University Rating')
plt.ylabel('Frequency')
plt.show(University_Rating_Dist)

#SOP
SOP_Dist = sns.histplot(df['SOP'])
plt.title('SOP Distribution')
plt.xlabel('SOP')
plt.ylabel('Frequency')
plt.show(SOP_Dist)


#LOR
LOR_Dist = sns.histplot(df['LOR '])
plt.title('LOR Distribution')
plt.xlabel('LOR')
plt.ylabel('Frequency')
plt.show(LOR_Dist)

#CGPA
CGPA_Dist = sns.histplot(df['CGPA'])
plt.title('CGPA Distribution')
plt.xlabel('CGPA')
plt.ylabel('Frequency')
plt.show(CGPA_Dist)

#Research
Research_Dist = sns.histplot(df['Research'])
plt.title('Research Distribution')
plt.xlabel('Research')
plt.ylabel('Frequency')
plt.show(Research_Dist)

#Chance of Admit
Chance_of_Admit_Dist = sns.histplot(df['Chance of Admit '])
plt.title('Chance of Admit Distribution')
plt.xlabel('Chance of Admit')
plt.ylabel('Frequency')
plt.show(Chance_of_Admit_Dist)
```
