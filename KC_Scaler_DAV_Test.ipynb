```python
IQ_after=np.array([113,127,89,70,127,104,69,127,115,99,104,120,95,129,106,71,94,112,154,96])
IQ_after
```

```python
mean_before = np.mean(IQ_before)
mean_before
```

```python
mean_after = np.mean(IQ_after)
mean_after
```

```python
std_before = np.std(IQ_before)
std_before
```

```python
std_after = np.std(IQ_after)
std_after
```

```python
confidence = 0.9
alpha = 0.1
```

```python
t_stat = (mean_before-mean_after)/(np.sqrt(((std_before**2)/20)+((std_after**2)/20)))
t_stat
```

```python
t_statistic, p_value = ttest_rel(IQ_after, IQ_before, alternative= "greater")
```

```python
print("T-STAT: ", t_statistic)
print("P-VAL: ", p_value)
```
